```python
# Filter for specific conditions and display selected columns
filtered_raw = raw[(raw["Passengers"] < 125) | (raw["Baggage_Complaints"] >= 8) | (raw["Delay_Minutes"] > 100)]
filtered_raw[["Date", "Passengers", "Delay_Minutes", "Baggage_Complaints"]]
```

```python
# Method 1: Remove rows containing missing values
df_drop = raw.dropna()

# Show descriptive statistics before removing missing values
print("DESCRIPTIVE STATISTICS BEFORE dropna()")
display(raw.describe().round(2))

# Show descriptive statistics after removing missing values
print("DESCRIPTIVE STATISTICS AFTER dropna()")
display(df_drop.describe().round(2))

# Compare the number of rows
print("Rows before:", len(raw))
print("Rows after:", len(df_drop))

# Calculate percentage of data lost
print("Data lost:", round((1 - len(df_drop) / len(raw)) * 100, 1), "%")

# Check missing values in each column after dropna()
print("Missing values after dropna():")
print(df_drop.isnull().sum())

# Count all missing cells in the dataset
print("Total blank cells:", df_drop.isnull().sum().sum())
```

**Method 2**

```python
# Create a copy of the original dataset
df_mean = raw.copy()

# Define the numeric columns to study
num_cols = ["Passengers", "Ticket_Price", "Delay_Minutes", "Flight_Duration_Min", "Fuel_Cost",
            "Ancillary_Revenue", "Baggage_Complaints", "Customer_Rating", "Crew_Overtime_Hours"]

# Display descriptive statistics BEFORE filling missing values
print("DESCRIPTIVE STATISTICS BEFORE IMPUTATION")
display(raw.describe().round(2))

# Fill missing values in numeric columns using the mean
df_mean[num_cols] = df_mean[num_cols].fillna(df_mean[num_cols].mean())

# Fill missing Route values using the most common route (mode)
df_mean["Route"] = df_mean["Route"].fillna(df_mean["Route"].mode()[0])

# Display descriptive statistics AFTER filling missing values
print("DESCRIPTIVE STATISTICS AFTER MEAN AND MODE IMPUTATION")
display(df_mean.describe().round(2))

# Compare the number of rows before and after
print("Rows before:", len(raw))
print("Rows after:", len(df_mean))

# Count missing cells remaining after imputation
print("Blanks left:", df_mean.isnull().sum().sum())
```

**Imputation with mode**

```python
df_mode = raw.copy()

for col in df_mode.columns:
    df_mode[col] = df_mode[col].fillna(df_mode[col].mode()[0])

print("Blanks left:", df_mode.isnull().sum().sum())

# Display the descriptive statistics of the mode-imputed dataset
print("DESCRIPTIVE STATISTICS AFTER MODE IMPUTATION")
display(df_mode.describe().round(2))
```

**Median Imputation**

```python
# Method 3: Fill with the median
df_median = raw.copy()

df_median[num_cols] = df_median[num_cols].fillna(
    df_median[num_cols].median()
)

# Route is text, so use the most common route
df_median["Route"] = df_median["Route"].fillna(
    df_median["Route"].mode()[0]
)

print("Blanks left:", df_median.isnull().sum().sum())
descriptive_stats = df_median.describe().round(2)
print("DESCRIPTIVE STATISTICS AFTER MEDIAN IMPUTATION")
display(descriptive_stats)
```

**Heatmaps After Treatment**

```python
# step 1 developing metrics

# A tiny example first
distance = np.array([1, 2, 3])
basket   = np.array([5, 10])

D, B = np.meshgrid(distance, basket)

print(D)
print(B)
```

```python
# step 2 dividing metrics

dist_values   = np.linspace(0.3, 4, 40)     # 40 distances, 0.3 to 4 km
basket_values = np.linspace(1, 12, 40)      # 40 basket sizes, 1 to 12 items

D, B = np.meshgrid(dist_values, basket_values)

print("Number of what-if cases:", D.size)
```

```python
# Step 3 Linear Regression - Prediction

# A simple formula from our data: delivery time from distance and basket size

from sklearn.linear_model import LinearRegression

formula = LinearRegression().fit(
    df[["Distance_KM", "Basket_Size"]],
    df["Delivery_Time_Min"]
)

print("Each extra km adds about", round(formula.coef_[0], 1), "minutes")
print("Each extra item adds about", round(formula.coef_[1], 2), "minutes")
```

```python
# Step 4 What if case

# Fill every what-if case with the expected delivery time
cases = pd.DataFrame({"Distance_KM": D.ravel(), "Basket_Size": B.ravel()})
T = formula.predict(cases).reshape(D.shape)
```

**Does staff count still matters once store size is held fix**

```python
m_staff_alone = smf.ols("Monthly_Sales_INR_Lakh ~ Staff_Count", data=df).fit()
m_staff_ctrl = smf.ols("Monthly_Sales_INR_Lakh ~ Staff_Count + Seats", data=df).fit()

print(m_staff_alone.summary().tables[1])
print("R-squared:", round(m_staff_alone.rsquared, 2))

print(m_staff_ctrl.summary().tables[1])
print("R-squared:", round(m_staff_ctrl.rsquared, 2))

print("Staff effect alone:", round(m_staff_alone.params["Staff_Count"], 2),
      "| p-value:", round(m_staff_alone.pvalues["Staff_Count"], 3))

print("Staff effect with Seats:", round(m_staff_ctrl.params["Staff_Count"], 2),
      "| p-value:", round(m_staff_ctrl.pvalues["Staff_Count"], 3))
```

**Interpretation** - A unit increase in staff count is associated with a ₹1.9441 lakh increase in monthly sales.

R-Squared – Staff count explains 9% of the variation in monthly sales.

P-value – 0.000. This means staff count has a statistically significant positive relationship with monthly sales.

**Does spends of social spending depends on how digital the store is:-**

```python
#step 1: Centralisation of variable
# Centre the numbers: subtract the average

df['Social_c'] = df['Social_Media_Spend_K'] - df['Social_Media_Spend_K'].mean()

df['App_c'] = df['App_Orders_Share_Pct'] - df['App_Orders_Share_Pct'].mean()

df['Wait_c'] = df['Avg_Wait_Time_Min'] - df['Avg_Wait_Time_Min'].mean()

df['Service_c'] = df['Service_Score'] - df['Service_Score'].mean()


# A yes or no flag: 1 = Metro store, 0 = other

df['Is_Metro'] = (df['City_Tier'] == 'Metro').astype(int)
```

**principal component analysis**

```python
groups = {
    "Customer Experience": ["Taste_Score", "Service_Score", "Cleanliness_Score", "Ambience_Score", "Staff_Friendliness_Score"],
    "Digital Engagement": ["App_Orders_Share_Pct", "Loyalty_Members_K", "App_Rating", "Social_Followers_K"],
    "Marketing Intensity": ["Social_Media_Spend_K", "Outdoor_Ad_Spend_K", "Influencer_Spend_K", "Promo_Offers_Count"],
    "Store Capacity": ["Store_Area_SqFt", "Seats", "Staff_Count"],
}

for name, cols in groups.items():
    c = df[cols].corr().values
    average = (c.sum() - len(cols)) / (len(cols) * (len(cols) - 1))  # average correlation inside the group
    print(f"{name}: {len(cols)} variables, average correlation {average:.2f}")
```

**PCA 1 Customer experience Index**

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

cols = groups["Customer Experience"]

# Step A: put every variable on the same scale
scaled = StandardScaler().fit_transform(df[cols])

# Step B: run PCA
pca = PCA()
scores = pca.fit_transform(scaled)

# Step C: how much does each component explain?
names = [f"PC{i+1}" for i in range(len(cols))]
print(pd.Series(pca.explained_variance_ratio_, index=names).round(2))
```

```python
#Step D # Step D: the loadings show how much each variable counts in PC1

sign = 1 if pca.components_[0].sum() > 0 else -1  # make sure a higher score means better

print(pd.Series(pca.components_[0] * sign, index=cols).round(2))
```

```python
# Step E: turn PC1 into an index from 0 to 100

raw_index = scores[:, 0] * sign

df["Experience_Index"] = (raw_index - raw_index.min()) / (raw_index.max() - raw_index.min()) * 100

variance_explained = {"Experience_Index": round(pca.explained_variance_ratio_[0] * 100)}

print(df[["Store_ID", "Experience_Index"]].sort_values("Experience_Index").head(5).round(1))
df.head()
```

**PCA 2- Digital engagement index**

```python
def make_index(columns, index_name):
    scaled = StandardScaler().fit_transform(df[columns])
    pca = PCA().fit(scaled)
    scores = pca.transform(scaled)[:, 0]

    sign = 1 if pca.components_[0].sum() > 0 else -1
    scores = scores * sign

    df[index_name] = (scores - scores.min()) / (scores.max() - scores.min()) * 100
    variance_explained[index_name] = round(pca.explained_variance_ratio_[0] * 100)

    print("Variance explained by PC1:", variance_explained[index_name], "%")
    print(pd.Series(pca.components_[0] * sign, index=columns).round(2))


make_index(groups["Digital Engagement"], "Digital_Index")
```
